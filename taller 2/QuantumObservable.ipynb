```python
class QuantumObservable:
    def __init__(self, observable_matrix, ket):
        self.observable = np.array(observable_matrix, dtype=complex)
        self.ket = np.array(ket, dtype=complex)
        
        # Verificar si la matriz es hermitiana
        if not np.allclose(self.observable, self.observable.conj().T):
            raise ValueError("La matriz no es hermitiana.")
    
    def expectation_value(self):
        # Calcula el valor esperado (media) del observable
        return np.vdot(self.ket, self.observable @ self.ket).real
    
    def variance(self):
        # Calcula la varianza del observable
        exp_value = self.expectation_value()
        observable_squared = self.observable @ self.observable
        exp_value_squared = np.vdot(self.ket, observable_squared @ self.ket).real
        return exp_value_squared - exp_value**2


# Ejemplo de uso:
observable = [[1, 0], [0, -1]]  # Matriz Pauli-Z
ket = [1/np.sqrt(2), 1/np.sqrt(2)]

observable_system = QuantumObservable(observable, ket)

print("Valor esperado:", observable_system.expectation_value())
print("Varianza:", observable_system.variance())
```
