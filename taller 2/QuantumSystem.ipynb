```python
import numpy as np

class QuantumSystem:
    def __init__(self, num_positions, ket_vector):
        self.num_positions = num_positions
        self.ket = np.array(ket_vector, dtype=complex)
        
        # Verificar si el ket es válido (normalizado)
        if not np.isclose(np.linalg.norm(self.ket), 1):
            raise ValueError("El vector ket no está normalizado.")
    
    def probability(self, position):
        # Calcula la probabilidad de encontrar la partícula en una posición particular
        if position < 0 or position >= self.num_positions:
            raise ValueError("La posición está fuera del rango.")
        amplitude = self.ket[position]
        return np.abs(amplitude)**2
    
    def transition_amplitude(self, ket_2):
        # Calcula la amplitud de transición entre dos kets
        ket_2 = np.array(ket_2, dtype=complex)
        return np.abs(np.vdot(self.ket, ket_2))**2  # Producto interno conjugado


# Ejemplo de uso:
# Supongamos un sistema de 3 posiciones con el ket [1/sqrt(2), 1/sqrt(2), 0]
system = QuantumSystem(3, [1/np.sqrt(2), 1/np.sqrt(2), 0])

# Probabilidad de encontrar la partícula en la posición 0
print(system.probability(0))  # Debería dar 0.5

# Probabilidad de transición a otro ket
ket_2 = [0, 1/np.sqrt(2), 1/np.sqrt(2)]
print(system.transition_amplitude(ket_2))
```
