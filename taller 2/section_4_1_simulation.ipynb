1. Simulación de un sistema con 3 posiciones


```python
from quantum_system import QuantumSystem

system = QuantumSystem(3, [1/np.sqrt(2), 1/np.sqrt(2), 0])
prob = system.probability(0)
print(f"Probabilidad de encontrar la partícula en la posición 0: {prob}")
```

2. Amplitud de transición entre dos vectores ket


```python
from quantum_system import QuantumSystem

system = QuantumSystem(3, [1/np.sqrt(2), 1/np.sqrt(2), 0])
ket_2 = [0, 1/np.sqrt(2), 1/np.sqrt(2)]
amplitude = system.transition_amplitude(ket_2)
print(f"Amplitud de transición: {amplitude}")
```

3. Cálculo de valores propios y transición a vectores propios


```python
from eigen_system import EigenSystem

observable = [[1, 0], [0, -1]]
ket = [1/np.sqrt(2), 1/np.sqrt(2)]

eigen_system = EigenSystem(observable)
probabilities = eigen_system.transition_probabilities(ket)
print(f"Probabilidades de transición a los vectores propios: {probabilities}")
```



4. Dinámica del sistema con matrices unitarias


```python
from quantum_dynamics import QuantumDynamics

U1 = [[0, 1], [1, 0]]  # Matriz X (Pauli-X)
U2 = [[1, 0], [0, 1]]  # Identidad

initial_state = [1, 0]
dynamics = QuantumDynamics(initial_state)
final_state = dynamics.evolve([U1, U2])
print(f"Estado final: {final_state}")
```
