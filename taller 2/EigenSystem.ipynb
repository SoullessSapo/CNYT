```python
class EigenSystem:
    def __init__(self, observable_matrix):
        self.observable = np.array(observable_matrix, dtype=complex)
        
        # Verificar si la matriz es hermitiana
        if not np.allclose(self.observable, self.observable.conj().T):
            raise ValueError("La matriz no es hermitiana.")
        
        # Calcular los valores propios y vectores propios
        self.eigenvalues, self.eigenvectors = np.linalg.eigh(self.observable)
    
    def transition_probabilities(self, ket):
        ket = np.array(ket, dtype=complex)
        probabilities = []
        
        for eigenvector in self.eigenvectors.T:
            prob = np.abs(np.vdot(ket, eigenvector))**2
            probabilities.append(prob)
        
        return probabilities

# Ejemplo de uso:
observable = [[1, 0], [0, -1]]  # Matriz Pauli-Z
ket = [1/np.sqrt(2), 1/np.sqrt(2)]

eigen_system = EigenSystem(observable)
print("Probabilidades de transición a los vectores propios:", eigen_system.transition_probabilities(ket))
```
