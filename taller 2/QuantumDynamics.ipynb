```python
class QuantumDynamics:
    def __init__(self, initial_state):
        self.ket = np.array(initial_state, dtype=complex)
    
    def evolve(self, unitary_matrices):
        # Aplicar una serie de matrices unitarias al ket
        for U in unitary_matrices:
            self.ket = U @ self.ket
        return self.ket

# Ejemplo de uso:
U1 = [[0, 1], [1, 0]]  # Matriz X (Pauli-X)
U2 = [[1, 0], [0, 1]]  # Identidad

initial_state = [1, 0]  # Ket inicial
dynamics = QuantumDynamics(initial_state)

final_state = dynamics.evolve([U1, U2])
print("Estado final:", final_state)
```
